```python
# Remove duplicate records if any

before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print("Records before removing duplicates:", before)
print("Records after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

```python
# Final missing-value check

missing = df.isnull().sum()

print("Total missing values:", missing.sum())

if missing.sum() == 0:
    print("Dataset is ready: no missing values found.")
else:
    print("\nColumns containing missing values:")
    print(missing[missing > 0])
```

```python
# Define features and target

target = "G3"

features = [
    "school",
    "sex",
    "age",
    "address",
    "famsize",
    "Pstatus",
    "Medu",
    "Fedu",
    "Mjob",
    "Fjob",
    "reason",
    "guardian",
    "traveltime",
    "studytime",
    "failures",
    "schoolsup",
    "famsup",
    "paid",
    "activities",
    "nursery",
    "higher",
    "internet",
    "romantic",
    "famrel",
    "freetime",
    "goout",
    "Dalc",
    "Walc",
    "health",
    "absences"
]

X = df[features]
y = df[target]

print("Number of features:", len(features))
print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Identify feature types

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    exclude=["object"]
).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
# Test preprocessing on training data

X_train_processed = preprocessor.fit_transform(X_train)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)
```

```python
# Save cleaned dataset for future use

cleaned_path = "../dataset/student_performance_cleaned.csv"

df.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print("Location:", cleaned_path)
```
