```python
import pandas as pd
import joblib

print("Prediction environment ready!")
```

```python
model = joblib.load(
    "../models/student_performance_model.pkl"
)

print("Student Performance Prediction Model loaded successfully!")
```

```python
df = pd.read_csv(
    "../dataset/student_performance_cleaned.csv"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
features = [
    "school",
    "sex",
    "age",
    "address",
    "famsize",
    "Pstatus",
    "Medu",
    "Fedu",
    "Mjob",
    "Fjob",
    "reason",
    "guardian",
    "traveltime",
    "studytime",
    "failures",
    "schoolsup",
    "famsup",
    "paid",
    "activities",
    "nursery",
    "higher",
    "internet",
    "romantic",
    "famrel",
    "freetime",
    "goout",
    "Dalc",
    "Walc",
    "health",
    "absences"
]
```

```python
sample_student = df[features].iloc[[0]]

prediction = model.predict(sample_student)[0]

print("Predicted Final Grade:", round(prediction, 2), "/ 20")
```

```python
def performance_category(score):
    if score < 8:
        return "At Risk"
    elif score < 12:
        return "Needs Improvement"
    elif score < 16:
        return "Good"
    else:
        return "Excellent"


category = performance_category(prediction)

print("=" * 40)
print("STUDENT PERFORMANCE PREDICTION")
print("=" * 40)
print("Predicted Grade :", round(prediction, 2), "/ 20")
print("Performance     :", category)
print("=" * 40)
```
